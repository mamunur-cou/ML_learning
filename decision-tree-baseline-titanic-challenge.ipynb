```python
import numpy as np 
import pandas as pd


import warnings
warnings.filterwarnings('ignore')
```

```python
train = pd.read_csv('/kaggle/input/titanic/train.csv')
test = pd.read_csv('/kaggle/input/titanic/test.csv')
full_data = [train,test]
```

```python
test_passenger_ids = test['PassengerId']
Dcol = ['PassengerId','Name', 'Ticket','Cabin']
for df in full_data:
    df.drop(Dcol,axis = 1,inplace = True)
```

```python
for df in full_data:
    med = df['Age'].median()
    df['Age'] = df['Age'].fillna(med)
    df['Embarked'] = df['Embarked'].fillna('S')
    
```

```python
test['Fare'] = test['Fare'].fillna(test['Fare'].median())
```

```python
from sklearn.preprocessing import StandardScaler
for df in full_data:
    ss =StandardScaler()
    df[['Age']]=ss.fit_transform(df[['Age']])
    df[['Fare']]= ss.fit_transform(df[['Fare']])
```

```python
for df in full_data:
    mp = {
        'male': 0,
        'female': 1
    }
    df['Sex'] = df['Sex'].map(mp)
```

```python
ecol = ['Embarked_C','Embarked_Q', 'Embarked_S']
train = pd.get_dummies(train,columns= ['Embarked'])
train[ecol] = train[ecol].astype(int)
```

```python
test = pd.get_dummies(test,columns= ['Embarked'])
test[ecol] = test[ecol].astype(int)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
y = train['Survived']
x = train.drop(['Survived'],axis = 1)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc = DecisionTreeClassifier(criterion = 'entropy',max_depth = 5,random_state = 42)
dtc.fit(x_train,y_train)
```

```python
test
```

```python
y_pred = dtc.predict(test)
```

```python
submission = pd.DataFrame(
    {
        'PassengerId':test_passenger_ids,
        'Survived': y_pred
    }
)
```

```python
submission.to_csv('submission.csv',index = False)
```
